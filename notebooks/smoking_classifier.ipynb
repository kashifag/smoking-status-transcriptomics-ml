```python
labels = []

parts = sample_line.strip().split("\t")[1:]

for p in parts:
    if "Never Smoker" in p:
        labels.append(0)
    elif "Smoker" in p:
        labels.append(1)

print("Samples:", len(labels))
print("Never smokers:", labels.count(0))
print("Smokers:", labels.count(1))
```

```python
import pandas as pd

expr = pd.read_csv(
    "GSE17913_series_matrix.txt",
    sep="\t",
    skiprows=64
)

expr.head()
```

```python
# Store gene IDs
gene_ids = expr["ID_REF"]

# Remove gene ID column
X = expr.drop(columns=["ID_REF"])

# Samples should be rows, genes should be columns
X = X.T

# Add labels
y = labels

print("X shape:", X.shape)
print("Number of labels:", len(y))
```

```python
# Convert everything to numeric
X = X.apply(pd.to_numeric, errors="coerce")

# Remove columns with too many missing values
X = X.dropna(axis=1, how="all")

print(X.shape)
```

```python
from sklearn.feature_selection import VarianceThreshold

selector = VarianceThreshold(threshold=0.1)

X_filtered = selector.fit_transform(X)

print("Original shape:", X.shape)
print("Filtered shape:", X_filtered.shape)
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

selector = SelectKBest(score_func=f_classif, k=500)

X_selected = selector.fit_transform(X, labels)

print("Selected shape:", X_selected.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

X_train, X_test, y_train, y_test = train_test_split(
    X_selected,
    labels,
    test_size=0.2,
    random_state=42,
    stratify=labels
)

model = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

model.fit(X_train, y_train)

predictions = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, predictions))
print(classification_report(y_test, predictions))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

ConfusionMatrixDisplay.from_estimator(
    model,
    X_test,
    y_test
)

plt.title("Smoking Status Classification")
plt.show()
```

```python
import pandas as pd

importance_df = pd.DataFrame({
    "Feature": range(len(model.feature_importances_)),
    "Importance": model.feature_importances_
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
)

importance_df.head(20)
```
